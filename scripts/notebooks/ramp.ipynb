# RAMP Mapping Visualization

This notebook visualizes the RAMP mapping over the input interval $[0, 1]$.

Adjustable mapping settings (from the RAMP component):
- `f_min`
- `f_max`
- `q`
- `convex_up`

```python
import numpy as np
import matplotlib.pyplot as plt


def ramp_mapping(x, f_min, f_max, q, convex_up):
    """Evaluate the RAMP mapping used in Neko-TOP."""
    q_eff = max(float(q), 1.0e-12)
    if convex_up:
        # convex-up form used by the implementation when convex_up = True
        return f_min + (f_max - f_min) * x * (q_eff + 1.0) / (q_eff + x)
    # standard RAMP form used when convex_up = False
    return f_min + (f_max - f_min) * x / (1.0 + q_eff * (1.0 - x))
```

```python
# Build a list of parameter sets as: (f_min, f_max, q)
parameter_sets = [
    (1000.0, 0.0, 0.1, True),
    (10000.0, 0.0, 0.1, True),
]

# Input interval fixed to [0, 1]
x = np.linspace(0.0, 1.0, 500)

plt.figure(figsize=(9, 5))
for i, (f_min, f_max, q, convex_up) in enumerate(parameter_sets, start=1):
    y = ramp_mapping(x, f_min=f_min, f_max=f_max, q=q, convex_up=convex_up)
    label = f'Set {i}: f_min={f_min}, f_max={f_max}, q={q}, convex_up={convex_up}'
    plt.plot(x, y, linewidth=2.2, label=label)

plt.xlim(0.0, 1.0)
plt.ylim(0.0, 1000.0)
plt.xlabel('X_in')
plt.ylabel('X_out')
plt.title(f'RAMP Mapping')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
