# RAMP and Borrvall & Petersson Mapping Visualization

This notebook visualizes the RAMP and Borrvall & Petersson mappings over
the input interval $[0, 1]$.

Adjustable mapping settings (shared by both mappings):
- `f_min` (the fluid-side value)
- `f_max` (the solid-side value)
- `q`

The two mappings share the shape but differ in assembly/direction:
- **RAMP** is increasing: $f(0)=f_{min}$, $f(1)=f_{max}$
  (x=0: fluid, x=1: solid).
- **Borrvall & Petersson** is decreasing: $f(0)=f_{max}$, $f(1)=f_{min}$
  (x=0: solid, x=1: fluid).

```python
import numpy as np
import matplotlib.pyplot as plt


def ramp_mapping(x, f_min, f_max, q):
    """Evaluate the standard RAMP mapping used in Neko-TOP.

    Increasing: f(0) = f_min (fluid), f(1) = f_max (solid).
    """
    q_eff = max(float(q), 1.0e-12)
    return f_min + (f_max - f_min) * x / (1.0 + q_eff * (1.0 - x))


def borrvall_petersson_mapping(x, f_min, f_max, q):
    """Evaluate the Borrvall & Petersson mapping used in Neko-TOP.

    Decreasing: f(0) = f_max (solid), f(1) = f_min (fluid).
    Uses the same shape function S(x) = x (q+1) / (x+q) as RAMP,
    assembled as f(x) = f_max + (f_min - f_max) * S(x).
    """
    q_eff = max(float(q), 1.0e-12)
    s = x * (q_eff + 1.0) / (x + q_eff)
    return f_max + (f_min - f_max) * s
```

```python
# Shared parameters (natural f_min < f_max ordering for both curves)
f_min = 0.0
f_max = 1000.0
q = 4.0

# Input interval fixed to [0, 1]
x = np.linspace(0.0, 1.0, 500)

y_ramp = ramp_mapping(x, f_min=f_min, f_max=f_max, q=q)
y_bp = borrvall_petersson_mapping(x, f_min=f_min, f_max=f_max, q=1 / q)

plt.figure(figsize=(9, 5))
plt.plot(x,
         y_ramp,
         linewidth=2.2,
         label=f'RAMP (increasing): f_min={f_min}, f_max={f_max}, q={q}')
plt.plot(x,
         y_bp,
         linewidth=2.2,
         label=f'Borrvall & Petersson (decreasing): '
         f'f_min={f_min}, f_max={f_max}, q={1/q}')

plt.xlim(0.0, 1.0)
plt.ylim(0.0, 1000.0)
plt.xlabel('X_in')
plt.ylabel('X_out')
plt.title('RAMP vs Borrvall & Petersson Mapping')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
